```python
##  Count the frequency of a particular character in provided string
## eg: 'how are you', here the frequency of h in this string is 2

s = input('enter the string')

term = input('what would you like to search for')

counter = 0

for i in s:
  if i == term:
    counter +=1

print('frequency', counter)

```

```python
## write a program which can remove a particular character from a string

s = input('enter the string')

term = input('what would you like to remove')
result = ''
for i in s:
  if i != term:
    result = result + i
print(result)


```

```python
## write a program to count the number of words in a string without split()

s = input('enter the string')
L= []
temp = ''

for i in s:
  if i != ' ':
    temp = temp +i
  else:
    L.append(temp)
    temp = ''
    
L.append(temp)
print(L)
```

```python
## write a python code to convert a string to title case without using Title()

s = input('enter the string')
L = []

for i in s.split():
  L.append(i[0].upper() + i[1:].lower())

print(" ".join(L))
  
```
