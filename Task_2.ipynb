### `Problem 1`: Write a program that will give you in hand monthly salary after deduction on CTC - HRA(10%), DA(5%), PF(3%) and taxes deduction as below:

> Salary(Lakhs) : Tax(%)

*   Below 5 : 0%
*   5-10 : 10%
*   10-20 : 20%
*   aboove 20 : 30%

````python
# Write code here
ctc = int(input('enter your Annual CTC:'))

if ctc < 5000000:
  salary = ctc*.82 ## here the dection is 18%(10,5,3) this is fix for all employee
if ctc < 1000000:
  salary = ctc*.72
if ctc < 2000000:
  salary = ctc*.62
else:
  salary = ctc*.52

print('your in hand salary will be-', round(salary/12,2))
````

### `Problem 2`: Write a program that take a user input of three angles and will find out whether it can form a triangle or not.

````python
# Write code here
first = int(input('enter the first angle'))
second = int(input('enter the second angle'))
third = int(input('enter the third angle'))

if first+second+third == 180 and first>0 and second>0 and third>0:
  print('form triangle')
else:
  print('doesnot form trangle')
````

### `Problem 3`: Write a program that will take user input of cost price and selling price and determines whether its a loss or a profit.

````python
# Write code here
cost_price = int(input('enter cost price'))
selling_price = int(input('enter selling price'))

if cost_price < selling_price:
  print('profit')
elif cost_price > selling_price:
  print('loss')
else:
  print('no loss no gain')
````

### `Problem 4`: Write a menu-driven program -
1. cm to fit

````python
# Write code here
menu = input("""
Hi select an option
1. cms to fit
2. km to miles
3. USD to INR
4. exit
""")

if menu =='1':
  cm = float(input('enter a cm value'))
  print('ft value is', 0.032*cm)
elif menu =='2':
  km = float(input('enter the km value'))
  print('miles value is', 0.62*km)
elif menu =='3':
  USD = float(input('enter the USD dollar'))
  print('indian rupees is', 80*USD)
else:
  exit()
````
