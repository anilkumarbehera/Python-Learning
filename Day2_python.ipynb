##### Operators in Python
1. Arithmetic Operator
2. Relational Operator
3. Logical Operator
4. Bitwise Operator
5. Assignment Operator
6. Membership Operators

```python
# 1. Arithmetic Operator
print(5+6)
print(6-5)
print(5*6)
print(6/5)
print(5//2) ## this called integer division(it will give you the integer as result)
print(5%2) ## this known as modulus operatr it gives us the remainder in output
print(5**2)
```

```python
#2. Relational operator
print(4>5)
print(4<5)
print(4>=4)
print(4<=4)
print(4==4)
print(4!=4)
```

```python
#3. Logical Operator
print(1 and 0)
print(1 or 0)
print(not 1)
```

##### Bitwise Operator


```python
## Bitwise AND
print(2 & 3)

# Bitwise OR
print(2 | 3)

# Bitwise xor
print(2 ^ 3)

print(~3) ## ~x = -(x+1)

print(4 >> 2)

print(4 << 2)
```

##### IF-ELSE Loop

```python
email = input('enter your email')
password = input('enter your password')

if email == 'anil@gmail.com' and password == '1234':
  print('welcome to my world')
elif email == 'anil@gmail.com' and password != '1234':
  print('Incorect password')
  password = input('enter password again')
  if password == '1234':
    print('Welcome to Login')
  else:
    print('Tum se na ho payega')
else:
  print('not correct ')
```

```python
## min of 3 numbers
num1  = int(input('first num'))
num2  = int(input('second num'))
num3  = int(input('third num'))
if num1 < num2 and num1 < num3:
  print('num1 is lowest', num1)
elif num2 < num3:
  print('num2 is lowest', num2)
else:
  print('num3 is lowest', num3)
```

```python
## menu driven calculator

fnum = int(input('enter the 1st number'))
snum = int(input('enter the second number'))

operation = input('Enter the operation')

if operation == '+':
  print(fnum + snum)
elif operation == '-':
  print(fnum - snum)
elif operation == '*':
  print(fnum * snum)
else:
  print('fnum / snum')
```

```python
## menu driven calculator

menu = input("""
Hi! How can i help you.
1. Enter 1 for pin check
2. enter 2 for balance check
3. enter 3 for withdrawl
4. enter 4 for cash deposit
5. press 5 for exit
""")

if menu == '1':
  print('pin change')
elif menu == '2':
  print('balance check')
elif menu == '3':
  print('withdrawl')
if menu == '4':
  print('cash deposit')
else:
  print('exit')
```

#### Loops in python
1. Need for Loops
2. while Loop
3. For Loop

```python
number = int(input('enter a number'))

i = 1
while i<11:
  print(number,'*', i, '=', number*i)
  i = i+1
```

```python
## while loop with else
x = 1
while x<3:
  print(x)
  x += 1

else:
  print('limit exceded')
```

```python
 ## guessing game
import random
jackpot = random.randint(1,100)

guess = int(input('number guess karo'))
counter = 1

while guess < jackpot:
	if guess < jackpot:
			print('galat! guess higher')
	else:
			print('galat! guess lower')
	
	guess = int(input('number guess karo'))
	counter =+1

else:
	print('correct guess')
	print('attempts', counter)
```
